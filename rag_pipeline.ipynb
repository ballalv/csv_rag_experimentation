```python
import pandas as pd

csv_path = "data/pharma_sales.csv"

df = pd.read_csv(csv_path)

print("Rows:", len(df))
print("Columns:", len(df.columns))

df.head()
```

```python
documents = []

for _, row in df.iterrows():
    document = {
        "title": row["Product"],
        "category": row["Business_Unit"],
        "content": " | ".join(f"{col}: {row[col]}" for col in df.columns),
        "source": "pharma_sales.csv"
    }

    documents.append(document)

print("Total documents:", len(documents))
print(documents[0])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

chunk_sizes = [100, 200, 300, 500, 1000]
results = []

for chunk_size in chunk_sizes:
    chunk_overlap = chunk_size // 10

    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap
    )

    all_chunks = []

    for doc in documents:
        chunks = text_splitter.split_text(doc["content"])
        all_chunks.extend(chunks)

    results.append({
        "Chunk Size": chunk_size,
        "Chunk Overlap": chunk_overlap,
        "Number of Documents": len(documents),
        "Number of Chunks": len(all_chunks)
    })

results
```

```python
import pandas as pd

results_df = pd.DataFrame(results)

results_df
```

```python
results_df.to_csv("results/chunk_experiment_results.csv", index=False)
print("Results saved successfully!")
```
